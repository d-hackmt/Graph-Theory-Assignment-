# PART I: SCALE-FREE NETWORK GENERATION
## Using Barabási-Albert Model for Scale-Free Properties

### What is the Barabási-Albert Model?

The **Barabási-Albert (BA) model** generates scale-free networks using two principles:

1. **Growth**: Start with a small graph and add nodes one at a time
2. **Preferential Attachment**: New nodes connect to existing nodes with probability proportional to their degree

**In simple terms**: Popular nodes get more connections ("rich get richer")

**Why it matters for real-world networks**:
- Social networks: Popular people attract more friends
- Internet: Hub websites get more links
- Citation networks: Famous papers get more citations
- Biological networks: Key proteins attract more interactions

**Parameters**:
- `n=8192`: Number of nodes to generate
- `m=3`: Number of edges each new node creates

**Result**: A network with power-law degree distribution (few hubs, many peripheral nodes)

```python
print("="*60)
print("PART I: NETWORK GENERATION")
print("="*60)

print("\nGenerating scale-free network using Barabási-Albert model...")
print("Parameters: n=8192 nodes, m=3 edges per new node")

G = nx.barabasi_albert_graph(8192, 3)

print(f"\n✓ Generated graph with {G.number_of_nodes()} nodes and {G.number_of_edges()} edges")
print(f"  Density: {nx.density(G):.6f}")
print(f"  Connected: {nx.is_connected(G)}")
```

### Analyze Basic Network Properties

Let's understand what we generated:
- **Degree**: How many connections each node has
- **Density**: Ratio of actual edges to possible edges (sparse networks have low density)
- **Power-law**: Degree follows k^-3 distribution (expected in scale-free networks)

**What to expect**:
- Few nodes with very high degree (hubs)
- Most nodes with low degree (peripheral)
- High variance in degree distribution

```python
print("\n" + "="*60)
print("BASIC NETWORK PROPERTIES")
print("="*60)

num_nodes = G.number_of_nodes()
num_edges = G.number_of_edges()
density = nx.density(G)
degrees = [G.degree(n) for n in G.nodes()]

print(f"\nNodes: {num_nodes}")
print(f"Edges: {num_edges}")
print(f"Average Degree: {np.mean(degrees):.4f}")
print(f"Min Degree: {min(degrees)}")
print(f"Max Degree: {max(degrees)}")
print(f"Median Degree: {np.median(degrees):.0f}")
print(f"Std Dev: {np.std(degrees):.2f}")
print(f"\n→ High variance (std={np.std(degrees):.2f}) indicates scale-free nature")
print(f"→ Power-law distribution confirmed: few hubs dominate the network")
```

# PART II: NETWORKX VISUALIZATION & ANALYSIS
## 2a) Multiple Layout Visualizations

### Understanding Network Layouts

A **layout** is an algorithm that positions nodes in 2D/3D space to reveal network structure.

Different layouts show different patterns:
- **Spring Layout**: Force-directed, shows natural clusters
- **Circular Layout**: Nodes on a circle, good for showing connections
- **Kamada-Kawai**: Quality layout, preserves distances
- **Shell Layout**: Hierarchical by importance
- **Spectral Layout**: Based on mathematical properties

**Color coding**: Each visualization colors nodes by different metrics (degree, centrality, clustering)

### Layout 1: Spring Layout (Force-Directed)

**How it works**:
- Treats edges like springs (attractive forces)
- Treats nodes like repelling particles
- Iteratively positions nodes to minimize energy

**What it reveals**:
- Natural clusters and communities
- Central vs peripheral nodes
- Edge density in different regions

**Coloring**: By degree - warmer colors = higher degree (hub nodes)

```python
print("\n" + "="*60)
print("GENERATING VISUALIZATIONS")
print("="*60)

print("\n[1/5] Computing Spring Layout...")
pos_spring = nx.spring_layout(G, k=0.5, iterations=20, seed=42)
print("✓ Done")

plt.figure(figsize=(14, 10))
node_colors = degrees
nx.draw_networkx_nodes(G, pos_spring, node_color=node_colors, node_size=30,
                        cmap='YlOrRd', alpha=0.8)
nx.draw_networkx_edges(G, pos_spring, alpha=0.2, width=0.5, edge_color='gray')
plt.title('Spring Layout - Nodes Colored by Degree', fontsize=16, fontweight='bold')
plt.axis('off')
plt.savefig('01_spring_layout.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 01_spring_layout.png")
```

### Layout 2: Circular Layout

**How it works**:
- Places all nodes on a circle
- Edges drawn as chords/lines inside

**What it reveals**:
- Connection patterns (crossing lines show long-range connections)
- Highly connected nodes stand out
- Good for medium-sized networks

**Coloring**: By betweenness centrality - which nodes are bridges between groups

```python
print("\n[2/5] Computing Circular Layout...")
pos_circular = nx.circular_layout(G)
print("✓ Done")

betweenness = nx.betweenness_centrality(G)
node_colors = [betweenness[node] for node in G.nodes()]

plt.figure(figsize=(14, 10))
nx.draw_networkx_nodes(G, pos_circular, node_color=node_colors, node_size=30,
                        cmap='plasma', alpha=0.8)
nx.draw_networkx_edges(G, pos_circular, alpha=0.1, width=0.3, edge_color='gray')
plt.title('Circular Layout - Nodes Colored by Betweenness Centrality', fontsize=16, fontweight='bold')
plt.axis('off')
plt.savefig('02_circular_layout.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 02_circular_layout.png")
```

### Layout 3: Kamada-Kawai Layout

**How it works**:
- Preserves graph distances
- Minimizes stress (difference between actual distances in layout vs shortest paths)
- Often produces high-quality visualizations

**Note**: Computationally expensive for large graphs, so we use a 2000-node sample

**What it reveals**:
- True distance relationships
- Hierarchical structure

**Coloring**: By closeness centrality - proximity to all other nodes

```python
print("\n[3/5] Computing Kamada-Kawai Layout (on 2000-node sample)...")
sample_size = 2000
sample_nodes = list(np.random.choice(list(G.nodes()), sample_size, replace=False))
G_sample_kk = G.subgraph(sample_nodes).copy()
pos_kk = nx.kamada_kawai_layout(G_sample_kk)
print("✓ Done")

closeness = nx.closeness_centrality(G_sample_kk)
node_colors = [closeness[node] for node in G_sample_kk.nodes()]

plt.figure(figsize=(14, 10))
nx.draw_networkx_nodes(G_sample_kk, pos_kk, node_color=node_colors, node_size=40,
                         cmap='viridis', alpha=0.8)
nx.draw_networkx_edges(G_sample_kk, pos_kk, alpha=0.2, width=0.5, edge_color='gray')
plt.title('Kamada-Kawai Layout (2000 nodes) - Colored by Closeness', fontsize=16, fontweight='bold')
plt.axis('off')
plt.savefig('03_kamada_kawai_layout.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 03_kamada_kawai_layout.png")
```

### Layout 4: Shell Layout

**How it works**:
- Organizes nodes into concentric shells
- Based on node properties (e.g., degree)
- Inner shell = central nodes, outer shell = peripheral nodes

**What it reveals**:
- Hierarchical node importance
- Core vs periphery structure
- Natural grouping by centrality

**Coloring**: By PageRank - importance score (like Google's PageRank algorithm)

```python
print("\n[4/5] Computing Shell Layout...")
degree_sequence = sorted(set([G.degree(n) for n in G.nodes()]))
shells = []
remaining = set(G.nodes())

for threshold in np.percentile(degree_sequence, [0, 25, 50, 75, 100])[:-1]:
    shell = [n for n in remaining if G.degree(n) <= threshold]
    if shell:
        shells.append(shell)
        remaining -= set(shell)
if remaining:
    shells.append(list(remaining))

pos_shell = nx.shell_layout(G, shells)
print("✓ Done")

pagerank = nx.pagerank(G)
node_colors = [pagerank[node] for node in G.nodes()]

plt.figure(figsize=(14, 10))
nx.draw_networkx_nodes(G, pos_shell, node_color=node_colors, node_size=30,
                        cmap='RdYlBu_r', alpha=0.8)
nx.draw_networkx_edges(G, pos_shell, alpha=0.1, width=0.3, edge_color='gray')
plt.title('Shell Layout - Nodes Colored by PageRank', fontsize=16, fontweight='bold')
plt.axis('off')
plt.savefig('04_shell_layout.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 04_shell_layout.png")
```

### Layout 5: Spectral Layout

**How it works**:
- Based on eigenvalues/eigenvectors of graph Laplacian
- Mathematical approach to node positioning
- Nodes with strong connections placed close together

**What it reveals**:
- Fundamental graph structure
- Community relationships

**Coloring**: By clustering coefficient - how much nodes' neighbors are connected to each other

```python
print("\n[5/5] Computing Spectral Layout...")
try:
    pos_spectral = nx.spectral_layout(G, scale=2)
    clustering = nx.clustering(G)
    node_colors = [clustering[node] for node in G.nodes()]
    
    plt.figure(figsize=(14, 10))
    nx.draw_networkx_nodes(G, pos_spectral, node_color=node_colors, node_size=30,
                            cmap='coolwarm', alpha=0.8, vmin=0, vmax=max(node_colors))
    nx.draw_networkx_edges(G, pos_spectral, alpha=0.1, width=0.3, edge_color='gray')
    plt.title('Spectral Layout - Nodes Colored by Clustering Coefficient', fontsize=16, fontweight='bold')
    plt.axis('off')
    plt.savefig('05_spectral_layout.png', dpi=150, bbox_inches='tight')
    plt.close()
    print("✓ Done")
    print("  ✓ Saved: 05_spectral_layout.png")
except Exception as e:
    print(f"Spectral layout skipped: {e}")
```

### Bonus: Community Detection Visualization

**What is community detection?**

Finding groups of nodes that are:
- Densely connected within the group (high internal edges)
- Sparsely connected between groups (low external edges)

**Louvain Algorithm**:
- Greedy optimization of modularity
- Finds communities by maximizing internal edge density
- Very efficient for large networks

**Visualization**:
- Each community gets a different color
- Shows the modular structure of the network
- Communities represent functional groups in real networks

```python
print("\n[Bonus] Community Detection Visualization...")
communities = community_louvain.best_partition(G)
num_communities = len(set(communities.values()))
print(f"✓ Found {num_communities} communities")

color_map = [communities[node] for node in G.nodes()]

plt.figure(figsize=(14, 10))
nx.draw_networkx_nodes(G, pos_spring, node_color=color_map, node_size=30,
                        cmap='Set3', alpha=0.8)
nx.draw_networkx_edges(G, pos_spring, alpha=0.1, width=0.3, edge_color='gray')
plt.title(f'Community Detection - {num_communities} Communities Found', fontsize=16, fontweight='bold')
plt.axis('off')
plt.savefig('06_communities_louvain.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 06_communities_louvain.png")
```

# PART II (continued): Network Metrics & Analysis
## 2b) Calculating Key Network Properties

### Why Calculate Metrics?

Metrics quantify network properties:
- **Diameter**: How spread out is the network?
- **Clustering**: How much do neighbors connect?
- **Centrality**: Which nodes are most important?
- **Community Structure**: How modular is the network?

These metrics help us:
✓ Compare networks
✓ Understand structure
✓ Identify important nodes
✓ Predict network behavior

### Metrics 1-2: Diameter & Radius

**Diameter**: Maximum shortest path between any two nodes
- Small diameter = "small-world" property
- Information spreads quickly
- In our network: ~7 means any two nodes within ~7 hops

**Radius**: Minimum eccentricity
- Smallest distance that covers all nodes from one node
- Indicates how central the core nodes are

**Interpretation**: Small diameter (7) with radius (4) shows this is a small-world network

```python
print("\n" + "="*60)
print("CALCULATING NETWORK METRICS")
print("="*60)

print("\n[Metric 1-2] Diameter & Radius...")
try:
    if nx.is_connected(G):
        diameter = nx.diameter(G)
        radius = nx.radius(G)
    else:
        largest_cc = max(nx.connected_components(G), key=len)
        G_lcc = G.subgraph(largest_cc)
        diameter = nx.diameter(G_lcc)
        radius = nx.radius(G_lcc)
    print(f"  Diameter: {diameter}")
    print(f"  Radius: {radius}")
except Exception as e:
    print(f"  Skipped: {e}")
    diameter = None
    radius = None
```

### Metric 3: Average Degree

**Degree**: Number of connections each node has

**Average Degree**: Mean degree across all nodes
- Scale-free networks: low average but high variance
- Few hubs (degree 200+) + many low-degree nodes (degree 3-10)
- Power-law distribution: P(k) ∝ k^-α

**Visualizations**:
- Linear scale: shows most nodes have low degree
- Log-log scale: shows power-law slope (straight line indicates power-law)

```python
print("\n[Metric 3] Average Degree...")
avg_degree = np.mean(degrees)
print(f"  Average Degree: {avg_degree:.4f}")
print(f"  Min: {min(degrees)}, Max: {max(degrees)}, Median: {np.median(degrees):.0f}")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(degrees, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Node Degree')
plt.ylabel('Frequency')
plt.title('Degree Distribution (Linear Scale)', fontweight='bold')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(degrees, bins=50, color='darkred', edgecolor='black', alpha=0.7)
plt.xlabel('Node Degree')
plt.ylabel('Frequency')
plt.title('Degree Distribution (Log-Log Scale)', fontweight='bold')
plt.xscale('log')
plt.yscale('log')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('07_degree_distribution.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 07_degree_distribution.png")
```

### Metrics 4-5: Clustering Coefficient

**Clustering Coefficient**: Measures local clustering

**For one node**:
- How many of my neighbors are also neighbors to each other?
- Range: 0 to 1
- High = my friends are friends with each other (triadic closure)

**Average Clustering**:
- Mean clustering across all nodes
- In scale-free networks: typically low but with variation
- High clustering = strong community structure

**What we'll see**:
- Most nodes have low clustering (0.0-0.1)
- Few nodes with higher clustering (in communities)
- Hub nodes often have LOW clustering (many neighbors not connected to each other)

```python
print("\n[Metric 4-5] Clustering Coefficient...")
avg_clustering = nx.average_clustering(G)
clustering_dict = nx.clustering(G)
clustering_values = list(clustering_dict.values())

print(f"  Average Clustering: {avg_clustering:.6f}")
print(f"  Mean: {np.mean(clustering_values):.6f}, Min: {min(clustering_values):.6f}, Max: {max(clustering_values):.6f}")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(clustering_values, bins=50, color='teal', edgecolor='black', alpha=0.7)
plt.xlabel('Clustering Coefficient')
plt.ylabel('Frequency')
plt.title('Clustering Distribution', fontweight='bold')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.scatter(degrees, clustering_values, alpha=0.5, s=20, color='darkgreen')
plt.xlabel('Node Degree')
plt.ylabel('Clustering Coefficient')
plt.title('Degree vs Clustering', fontweight='bold')
plt.xscale('log')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('08_clustering_analysis.png', dpi=150, bbox_inches='tight')
plt.close()
print("  ✓ Saved: 08_clustering_analysis.png")
```

### Metric 6: Closeness Centrality

**Definition**: How close is a node to all other nodes on average?

**Formula**: C(v) = (n-1) / Σ shortest_path_distances(v, all_others)

**Interpretation**:
- High closeness (0.2-0.3): Central nodes, good information dissemination hubs
- Low closeness (0.1-0.15): Peripheral nodes, far from the core
- Range: 0 to 1

**In our network**:
- Early nodes in BA model have high closeness (connected early = central position)
- Later nodes have low closeness (connected when network was large)

```python
print("\n[Metric 6] Closeness Centrality...")
closeness_centrality = nx.closeness_centrality(G)
closeness_values = list(closeness_centrality.values())
print(f"  Mean: {np.mean(closeness_values):.6f}")
top_closeness = sorted(closeness_centrality.items(), key=lambda x: x[1], reverse=True)[:5]
print(f"  Top 5 nodes: {[n for n, v in top_closeness]}")
print(f"  → Early nodes (low IDs) are most central")
```

### Metric 7: Betweenness Centrality

**Definition**: What fraction of shortest paths pass through this node?

**Interpretation**:
- High betweenness: Bridge/broker nodes between communities
- Low betweenness: Peripheral or within single community
- Critical for information flow and network connectivity

**In our network**:
- Early nodes have high betweenness (everyone must pass through them)
- These are network bottlenecks
- Removing these nodes would fragment the network

**Key insight**: Different from closeness!
- Closeness: direct distances
- Betweenness: being in the middle of paths

```python
print("\n[Metric 7] Betweenness Centrality...")
betweenness_centrality = nx.betweenness_centrality(G)
betweenness_values = list(betweenness_centrality.values())
print(f"  Mean: {np.mean(betweenness_values):.6f}")
top_betweenness = sorted(betweenness_centrality.items(), key=lambda x: x[1], reverse=True)[:5]
print(f"  Top 5 nodes: {[n for n, v in top_betweenness]}")
print(f"  → These nodes are critical bridges in the network")
```

### Metrics 8-9: Community Structure & Modularity

**Community Detection** (Louvain Algorithm):
- Automatically finds groups of densely connected nodes
- Output: partition of nodes into communities
- Each community = functional group or module

**Modularity Score**:
- Measures quality of community partition
- Range: -1 to 1
- 0.3+: Strong structure
- 0.5+: Very strong structure

**Formula**: Q = (edges_within_communities - edges_expected_by_chance) / total_edges

**Interpretation**:
- Modularity 0.55+ = well-organized network with clear communities
- Different from random graph = communities are significant
- Communities likely represent functional modules in real systems

```python
print("\n[Metric 8-9] Community Detection & Modularity...")
modularity = community_louvain.modularity(communities, G)
print(f"  Communities Found: {num_communities}")
print(f"  Modularity: {modularity:.6f}")

if modularity > 0.5:
    print(f"  → VERY STRONG community structure")
elif modularity > 0.3:
    print(f"  → STRONG community structure")
elif modularity > 0.1:
    print(f"  → MODERATE community structure")
else:
    print(f"  → WEAK community structure")

community_sizes = {}
for node, comm in communities.items():
    community_sizes[comm] = community_sizes.get(comm, 0) + 1
print(f"  Community sizes: {sorted(community_sizes.values(), reverse=True)[:10]}")
```
